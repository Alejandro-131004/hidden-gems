```python
import pandas as pd
from compare import compare
```

```python
df1 = pd.read_csv('../data/sample500_bundesliga2.csv')
df2 = pd.read_csv('../fetch/out/Germany_2_Bundesliga__2025_2026.csv')
```

```python
SORT_BY = ["Market value", "Team within selected timeframe", "Team", "Player"]

def order(df):
    d = df.copy()
    d["_mv"] = pd.to_numeric(d["Market value"], errors="coerce")   # handles int or "€10M"
    d = d.sort_values(
        ["_mv", "Team within selected timeframe", "Team", "Player"],
        ascending=[False, True, True, True],   # market value high→low, rest A→Z
        na_position="last",
        kind="mergesort",                      # stable
    ).drop(columns="_mv").reset_index(drop=True)
    return d
```

```python
df1 = order(df1)
df2 = order(df2)
```
